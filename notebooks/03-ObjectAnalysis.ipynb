```python
from pathlib import Path
from ultralytics import YOLO
import cv2
import os
```

```python
folder_path = Path('../data/cropped/')

image_extensions = {'.jpg', '.jpeg', '.png', '.gif', '.bmp', '.webp'}

image_lst = [
    file for file in folder_path.iterdir() 
    if file.is_file() and file.suffix.lower() in image_extensions
]

print(image_lst)
```

```python
os.makedirs('../results/Images/Object_Detection', exist_ok=True)
os.makedirs('../results/models', exist_ok=True)

def run_object_detection(img_path):
    model = YOLO('../results/models/yolov8n.pt')
    results = model(img_path)
    return results[0]
```

```python
save_path = '../results/Images/Object_Detection'

for image_path in image_lst:
    res_plotted = run_object_detection(str(image_path))

    cv2.imwrite(os.path.join(save_path, image_path.name), res_plotted.plot())

print ("finish")
```
