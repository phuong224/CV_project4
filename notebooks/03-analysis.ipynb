```python
from pathlib import Path

folder_path = Path('../data/raw/')

image_extensions = {'.jpg', '.jpeg', '.png', '.gif', '.bmp', '.webp'}

image_path_objects = [
    file for file in folder_path.iterdir() 
    if file.is_file() and file.suffix.lower() in image_extensions
]

image_list = [str(path) for path in image_path_objects]

print(f"Đã tìm thấy {len(image_list)} ảnh.")
print(image_list)
```

```python
# Tạo thư mục lưu trữ nếu chưa có
os.makedirs('../results/Images/Object_Detection', exist_ok=True)
```

```python
from ultralytics import YOLO
import cv2
import os


def run_object_detection(img_path):
    model = YOLO('yolov8n.pt')
    
    results = model(img_path)

    return results[0]
    
   
```

```python
save_path = '../results/Images/Object_Detection'

for image_path in image_list:
    # Dùng os.path.basename để lấy tên file từ chuỗi
    ten_file = os.path.basename(image_path)
    
    # Ghép đường dẫn với tên file
    final_destination = os.path.join(save_path, ten_file)
    
    # Chạy thử với ảnh gốc của bạn
    res_plotted = run_object_detection(image_path)

    cv2.imwrite(final_destination, res_plotted.plot())
    print(f"Đã lưu kết quả nhận diện vào: {final_destination}")
```
